```python
def subtract(a, b):
    return a - b
```

```python
subtract(5, 2)
```

```python
subtract(8, 1)
```

```python
subtract(9, 4)
```

```python
def distance_from_zero(x):
    if x < 0:
        return subtract(0, x)
    else:
        return x
```

```python
distance_from_zero(-5)
```

```python
distance_from_zero(10)
```

```python
distance_from_zero(-400)
```
