```python
# defining the function
def add(a,b):
    result = a+b
    return result
```

```python
# 2 example calls
print(add(5, 3))  # Output: 8
print(add('hello', 'world'))  # Output: 'helloworld'
```

```python
# defining the function extension
def sum_list(values):
    listsum = 0
    for i in values:
        listsum = add(listsum, i)
    return listsum
```

```python
# example calls
print(sum_list([1, 2, 3, 10])) # output: 16
print(sum_list([5, 5, 10, 9])) # output: 29
```
